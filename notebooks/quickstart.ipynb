# Quick start

This notebook reproduces a few headline results interactively. Run `pip install -e .` from the repository root first. Conventions: signature (-,+,+,+), G = c = 1; see `docs/mathematical_conventions.md`.

```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
from beyond_spacetime import symbolic as S, geodesics as G, tides as T, regular as H, holography as Hh, perturbations as P
```

## 1. Curvature of Schwarzschild from the metric

The Kretschmann scalar is computed symbolically from the metric; nothing is assumed.

```python
sch = S.schwarzschild()
print('Ricci tensor vanishes:', sch.ricci() == sp.zeros(4, 4))
print('Kretschmann scalar    :', sch.kretschmann())
```

## 2. Proper time left after crossing the horizon

Analytic maximum is pi M at E = 0; for E = 1 it is 4M/3.

```python
print(G.tau_inside(0.0) / np.pi, G.tau_inside(1.0), 4/3)
```

## 3. Tidal tensor from the full Riemann tensor

Independent of the conserved energy E.

```python
for r, E in [(3.0, 1.3), (0.7, 0.4)]:
    print(r, E, np.diag(T.frame_tidal_tensor(r, E)), np.diag(T.tidal_closed_form(r)))
```

## 4. Regular black hole: curvature stays finite

Core value is 24 / l^4.

```python
l = 0.5
print(H.kretschmann(1e-4, 1.0, l), H.kretschmann_core_limit(l))
print('horizons:', H.horizons(1.0, l))
```

## 5. Holographic entanglement entropy with the homology constraint

```python
rp, rc = 1.5, 1e5
ds = Hh.transition_angle(rp)
print('transition angle:', ds)
print('S_A - S_Ac - S_BH at dphi=6:', Hh.rt_entropy(6.0, rp, rc) - Hh.rt_entropy(2*np.pi - 6.0, rp, rc) - Hh.bh_entropy(rp))
```

## 6. Quasinormal mode from Leaver's continued fraction

```python
print(P.leaver_qnm(2, 2, 0), '(literature: 0.37367 - 0.08896j)')
```
